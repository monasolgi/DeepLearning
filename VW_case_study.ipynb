```python
df=df[df["road_segment_id"] !=21894 ].copy()
```

```python
weather_check=weather.groupby("timestamp")[["tempC", "windspeedKmph", "precipMM"]].nunique() # number of unique values in each column
weather_check.head()
```

```python
weather_check[weather_check["tempC"] > 1]
```

```python
temp_median=weather.groupby("timestamp")["tempC"].median()
df["tempC"]=df["tempC"].fillna( df["timestamp"].map(temp_median))
```

```python
wind_temp=weather.groupby("timestamp")["windspeedKmph"].median()
df["windspeedKmph"]=df["windspeedKmph"].fillna(df["timestamp"].map(wind_temp))
```

```python
precip_median = weather.groupby("timestamp")["precipMM"].median()

df["precipMM"] = df["precipMM"].fillna(
    df["timestamp"].map(precip_median))
```
